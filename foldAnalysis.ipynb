```python
import pandas as pd
import numpy as np
import os
```

```python
def load_data(fold_num):
    return pd.read_csv(f"bcg_dataset/feat_fold_{fold_num}.csv")
```

```python
stats = []
for fold in range(0, 5):
    df = load_data(fold)
    for col in ['SP', 'DP']:
        s = df[col]
        stats.append({
            'fold': fold,
            'target': col,
            'mean': s.mean(),
            'std': s.std(),
            'min': s.min(),
            'max': s.max(),
            'len': len(s)
        })

results = pd.DataFrame(stats).set_index(['fold', 'target'])
print(results.to_string())
```
