```python
import pandas as pd
import numpy as np
import pickle
from scipy.stats import spearmanr
from pathlib import Path

def load_data(fold):
    return pd.read_csv(f"bcg_dataset/feat_fold_{fold}.csv")

all_data = pd.concat([load_data(i) for i in range(5)])
sbp = all_data.loc[:,"SP"].values
dbp = all_data.loc[:,"DP"].values
all_features = all_data.columns[4:]

sbp_ranking_monotone = []
dbp_ranking_monotone = []

for feat in all_features:
    r_sbp, _ = spearmanr(all_data[feat].values, sbp)
    r_dbp, _ = spearmanr(all_data[feat].values, dbp)
    sbp_ranking_monotone.append((feat, r_sbp))
    dbp_ranking_monotone.append((feat, r_dbp))

sbp_ranking_monotone.sort(key=lambda x: abs(x[1]), reverse=True)
dbp_ranking_monotone.sort(key=lambda x: abs(x[1]), reverse=True)

with open("results_RF_importances/rf_results.pkl", "rb") as f:
    results = pickle.load(f)

results.update({
    "sbp_ranking_monotone": sbp_ranking_monotone,
    "dbp_ranking_monotone": dbp_ranking_monotone
})

with open("results_RF_importances/rf_results.pkl", "wb") as f:
    pickle.dump(results, f)

print("\nSaved sbp_ranking_monotone and dbp_ranking_monotone to pkl")
```
